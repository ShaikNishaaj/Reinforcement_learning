Assignment1: Calculation of State and Action Values Using Bellman Expectation Equations for
stochastic policy and deterministic environment

```python
 # Assignment 1: Bellman Expectation Equations
# Calculate state values and action values algebraically.

gamma = 0.9

# Policy probabilities
p_s0_down = 0.9
p_s0_right = 0.1
p_s1_right = 0.9
p_s1_up = 0.1

# Rewards
normal_reward = -1
wall_reward = -1
goal_reward = 100
v_goal = 0
v_wall = 0  # Assumption: wall-hit transition ends here.

# Bellman equations:
# V(S0) = -1 + 0.81 * V(S1)
# V(S1) = 89.9 + 0.09 * V(S0)

# Solve the simultaneous equations algebraically.
v_s0 = (
    -1 + (gamma ** 2) * 89.9
) / (
    1 - (gamma ** 2) * 0.09
)

v_s1 = 89.9 + 0.09 * v_s0

# Calculate action values.
q_s0_down = normal_reward + gamma * v_s1
q_s0_right = wall_reward + gamma * v_wall

q_s1_right = goal_reward + gamma * v_goal
q_s1_up = normal_reward + gamma * v_s0

# Verify the state values using the policy probabilities.
verified_v_s0 = (
    p_s0_down * q_s0_down
    + p_s0_right * q_s0_right
)

verified_v_s1 = (
    p_s1_right * q_s1_right
    + p_s1_up * q_s1_up
)

# Display results.
print("STATE VALUES")
print(f"V(S0) = {v_s0:.2f}")
print(f"V(S1) = {v_s1:.2f}")

print("\nACTION VALUES")
print(f"Q(S0, Down)  = {q_s0_down:.2f}")
print(f"Q(S0, Right) = {q_s0_right:.2f}")
print(f"Q(S1, Right) = {q_s1_right:.2f}")
print(f"Q(S1, Up)    = {q_s1_up:.2f}")

print("\nVERIFICATION")
print(f"Verified V(S0) = {verified_v_s0:.2f}")
print(f"Verified V(S1) = {verified_v_s1:.2f}")

# Confirm that the equations are satisfied.
assert abs(v_s0 - verified_v_s0) < 1e-9
assert abs(v_s1 - verified_v_s1) < 1e-9

print("\nVerification successful!")
```

Assignment 2: Calculation of State and Action Values Using Bellman Expectation Equations for
stochastic environment and deterministic policy

```python

# Assignment 2: Bellman Expectation Equations
# Iterative Policy Evaluation and Action Value Calculation

# Parameters
gamma = 0.9
theta = 0.000001

# Define the policy
policy = {
    "S0": "Down",
    "S1": "Right"
}

# Define all state-action transitions
# Each transition contains: (next_state, probability, reward)

transitions = {
    ("S0", "Down"): [
        ("Wall", 0.1, -1),
        ("S1", 0.9, -1)
    ],

    ("S0", "Right"): [
        ("Wall", 1.0, -1)
    ],

    ("S1", "Right"): [
        ("G", 0.9, 100),
        ("S0", 0.1, -1)
    ],

    ("S1", "Up"): [
        ("S0", 1.0, -1)
    ]
}

# Initialize state values
V = {
    "S0": 0.0,
    "S1": 0.0
}

# Terminal states
terminal_states = {"Wall", "G"}

# Iterative Policy Evaluation
iteration = 0

while True:
    new_V = {}

    # Calculate new values using previous iteration values
    for state in ["S0", "S1"]:
        action = policy[state]
        value = 0.0

        for next_state, probability, reward in transitions[
            (state, action)
        ]:
            if next_state in terminal_states:
                future_value = 0.0
            else:
                future_value = V[next_state]

            value += probability * (
                reward + gamma * future_value
            )

        new_V[state] = value

    # Calculate maximum change
    delta = max(
        abs(new_V["S0"] - V["S0"]),
        abs(new_V["S1"] - V["S1"])
    )

    V = new_V
    iteration += 1

    # Stop when convergence is reached
    if delta < theta:
        break

# Function to calculate action values
def calculate_q(state, action):
    q_value = 0.0

    for next_state, probability, reward in transitions[
        (state, action)
    ]:
        if next_state in terminal_states:
            future_value = 0.0
        else:
            future_value = V[next_state]

        q_value += probability * (
            reward + gamma * future_value
        )

    return q_value


# Calculate all action values
Q = {}

for state, action in transitions:
    Q[(state, action)] = calculate_q(state, action)


# Display results
print("ITERATIVE POLICY EVALUATION")
print("--------------------------------")
print("Number of iterations:", iteration)
print("Final delta:", format(delta, ".10f"))

print("\nCONVERGED STATE VALUES")
print("--------------------------------")
print("V_pi(S0) =", format(V["S0"], ".10f"))
print("V_pi(S1) =", format(V["S1"], ".10f"))

print("\nACTION VALUES")
print("--------------------------------")
print("Q_pi(S0, Down)  =", format(Q[("S0", "Down")], ".10f"))
print("Q_pi(S0, Right) =", format(Q[("S0", "Right")], ".10f"))
print("Q_pi(S1, Right) =", format(Q[("S1", "Right")], ".10f"))
print("Q_pi(S1, Up)    =", format(Q[("S1", "Up")], ".10f"))

print("\nPOLICY CONSISTENCY CHECK")
print("--------------------------------")
print(
    "S0:",
    abs(V["S0"] - Q[("S0", "Down")]) < theta
)
print(
    "S1:",
    abs(V["S1"] - Q[("S1", "Right")]) < theta
)
```
