```python

GAMMA = 0.9
GOAL_REWARD = 10
THETA = 1e-6

states = ["S0", "S1"]
actions = ["Up", "Down", "Left", "Right"]

# State transitions: next state and reward
transitions = {
    "S0": {
        "Up": ("S0", -1),
        "Down": ("S1", -1),
        "Left": ("S0", -1),
        "Right": ("S0", -1)
    },
    "S1": {
        "Up": ("S0", -1),
        "Down": ("S1", -1),
        "Left": ("S1", -1),
        "Right": ("G", GOAL_REWARD)
    }
}

# 1. Display MDP
print("STATES, ACTIONS AND TRANSITIONS")
for s in states:
    for a in actions:
        ns, r = transitions[s][a]
        print(s, a, "->", ns, "Reward:", r)

# 2. Initialize state values
V = {"S0": 0.0, "S1": 0.0, "G": 0.0}

print("\nINITIAL VALUES:", V)

# 3. Value Iteration
iteration = 0
history = []

while True:
    old_V = V.copy()
    new_V = V.copy()

    for s in states:
        q_values = []

        for a in actions:
            ns, reward = transitions[s][a]
            q = reward + GAMMA * old_V[ns]
            q_values.append(q)

        new_V[s] = max(q_values)

    delta = max(abs(new_V[s] - old_V[s]) for s in states)

    V = new_V
    iteration += 1

    history.append((iteration, V["S0"], V["S1"], delta))

    if delta < THETA:
        break

# 4. Display convergence table
print("\nVALUE ITERATION RESULTS")
print("Iteration | V(S0) | V(S1) | Delta")

print("0 | 0.00 | 0.00 | -")

for row in history:
    print(row)

# 5. Calculate action values
Q = {}

for s in states:
    Q[s] = {}

    for a in actions:
        ns, reward = transitions[s][a]
        Q[s][a] = reward + GAMMA * V[ns]

# 6. Display Q-values
print("\nACTION VALUE TABLE")

for s in states:
    print(s, Q[s])

# 7. Extract optimal policy
print("\nOPTIMAL POLICY")

for s in states:
    best_value = max(Q[s].values())

    best_actions = [
        a for a in actions
        if abs(Q[s][a] - best_value) < 1e-9
    ]

    print(s, "->", best_actions)

# 8. Bellman Optimality Verification
print("\nBELLMAN VERIFICATION")

for s in states:
    max_q = max(Q[s].values())

    print(
        f"V*({s}) = {V[s]:.2f}, "
        f"Max Q*({s},a) = {max_q:.2f}"
    )

    assert abs(V[s] - max_q) < 1e-9

print("\nAll Bellman checks passed!")
```
