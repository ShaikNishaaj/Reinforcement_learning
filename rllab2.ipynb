```python
import random

machines = ['A', 'B', 'C']
epsilon = 0.2
plays = 10

reward_pattern = {
    'A': [3, 4, 2, 5],
    'B': [6, 7, 5, 8],
    'C': [4, 3, 5, 4]
}

N = {'A': 0, 'B': 0, 'C': 0}
R = {'A': [], 'B': [], 'C': []}
Q = {'A': 0, 'B': 0, 'C': 0}

for i in range(plays):

    r = random.random()

    if r < epsilon:
        machine = random.choice(machines)
        method = "Exploration"
    else:
        machine = max(Q, key=Q.get)
        method = "Exploitation"
    reward = reward_pattern[machine][N[machine] % len(reward_pattern[machine])]

    print("\nPlay", i + 1)
    print("Selected Machine:", machine)
    print("Method:", method)
    print("Reward:", reward)

    N[machine] += 1
    R[machine].append(reward)

    Q[machine] = Q[machine] + (reward - Q[machine]) / N[machine]

print("\nFinal Results")
print("----------------")

for machine in machines:
    print("Machine", machine)
    print("N =", N[machine])
    print("Rewards =", R[machine])
    print("Q =", Q[machine])

print("\nGreedy Selected Machine:", max(Q, key=Q.get))
```
