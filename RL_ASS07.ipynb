```python
# Assignment 1: Calculate Undiscounted and Discounted Returns

def calculate_returns():
    # Accept number of time steps
    n = int(input("Enter number of time steps: "))

    # Accept rewards
    rewards = []
    for i in range(n):
        r = float(input(f"Enter reward for time step {i+1}: "))
        rewards.append(r)

    # Undiscounted return
    undiscounted_return = sum(rewards)

    # Accept discount factor
    gamma = float(input("Enter discount factor (γ): "))

    # Discounted return
    discounted_return = sum([rewards[t] * (gamma ** t) for t in range(n)])

    # Display results
    print("\nEntered Rewards:", rewards)
    print("Undiscounted Return:", undiscounted_return)
    print("Discounted Return:", discounted_return)

# Run the function
calculate_returns()
```

```python
# grid_env_det.py - Deterministic Grid World MDP

class GridWorldEnv:
    def __init__(self):
        # Define state space
        self.states = ["S0", "S1", "S3", "S4", "G"]

        # Define action space
        self.actions = ["up", "down", "left", "right"]

        # Define rewards
        self.rewards = {
            "normal": -1,
            "goal": 10,
            "obstacle": -5
        }

        # Discount factor
        self.gamma = 0.9

        # Transition function (deterministic)
        self.transitions = {
            "S0": {"up":"S0", "down":"S0", "left":"S0", "right":"S1"},
            "S1": {"up":"S1", "down":"S3", "left":"S0", "right":"G"},
            "S3": {"up":"S1", "down":"S3", "left":"S3", "right":"S4"},
            "S4": {"up":"G", "down":"S4", "left":"S3", "right":"S4"},
            "G":  {"up":"G", "down":"G", "left":"G", "right":"G"} # terminal
        }

    def step(self, state, action):
        """Return next state and reward given current state and action"""
        if state == "G":
            return state, self.rewards["goal"], True

        next_state = self.transitions[state][action]

        # Reward logic
        if next_state == "G":
            reward = self.rewards["goal"]
            done = True
        else:
            reward = self.rewards["normal"]
            done = False

        return next_state, reward, done


# Example usage
if __name__ == "__main__":
    env = GridWorldEnv()
    state = "S0"
    print("Starting at:", state)

    actions = ["right", "right"]  # Move S0 -> S1 -> G
    for a in actions:
        next_state, reward, done = env.step(state, a)
        print(f"Action: {a}, Next State: {next_state}, Reward: {reward}, Done: {done}")
        state = next_state
        if done:
            break
```
